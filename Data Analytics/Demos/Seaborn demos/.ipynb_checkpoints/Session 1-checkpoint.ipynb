```python
import seaborn as sns
```

```python
data = sns.load_dataset('tips')
data.head()
```

```python
import matplotlib.pyplot as plt
```

```python
sns.barplot(data = data, x = 'sex', y = 'tip')
plt.show()
```

```python
sns.histplot(data)
```

```python
sns.lineplot(data, x='total_bill', y='tip')
```

```python
sns.lineplot(data, x='total_bill', y='tip', hue='sex')
```

```python
sns.scatterplot(data, x='total_bill', y='tip', hue='sex')
```
