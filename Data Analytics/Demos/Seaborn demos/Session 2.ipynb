### Facetgrid
#### 1. Same graph will displayed with categories
#### 2. Create multiple graphs based on subset of data

```python
g = sns.FacetGrid(data, col='Species')
plt.show()
```

```python
g = sns.FacetGrid(data, col='Species')
g.map(sns.scatterplot, 'Petal_Length', 'Petal_Width')
plt.show()
```

### Pairplot
#### 1. Relationship between numerical columns
#### 2. Gives distribution for diagonal
#### 3. Scatterplot for relationship between other columns

```python
sns.pairplot(data)
```

```python
sns.pairplot(data, hue = 'Species')
```

### Geographical plot
#### 1. Visualize on map

```python
data = pd.DataFrame({
    'city':['Pune', 'Mumbai', 'Delhi'],
    'lat':[18.5204, 19.0760, 28.7041],
    'lon':[73.8567, 72.8777, 77.1025],
    'sale':[200, 500, 350]})
data
```

```python
import plotly.express as ex
# import plotly.io as pio
```

```python
# pio.renderers.default = 'notebook'
ex.scatter_geo(data, lat='lat', lon='lon', size='sale', hover_name='city')
```

```python
fig = ex.scatter_geo(data, lat='lat', lon='lon', size='sale', hover_name='city')
fig.update_layout(template='plotly_dark')
fig.show()
```

### Subplot 
#### 1. Used to add multiple graphs in 1

```python
data = pd.read_csv("D:/Dataset/Iris.csv")
data.head()
plt.subplot(3, 2, 1)
sns.scatterplot(data, x='Petal_Width', y='Petal_Length', hue = 'Species', legend=False)

plt.subplot(3, 2, 2)
sns.lineplot(data, x='Petal_Width', y='Petal_Length', hue = 'Species', legend=False)

plt.subplot(3, 2, 3)
sns.boxplot(data['Petal_Length'])

plt.subplot(3, 2, 4)
sns.scatterplot(data, x='Petal_Width', hue = 'Species')

plt.subplot(3, 2, 5)
sns.scatterplot(data, x='Sepal_Length', y='Sepal_Width')
```
