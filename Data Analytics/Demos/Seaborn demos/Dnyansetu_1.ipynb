```python
import seaborn as s
import matplotlib.pyplot as p
```

```python
months = ['Jan', 'Feb', 'Mar', 'Apr']
sales = [100, 150, 120, 200]
```

```python
s.lineplot(x = months, y = sales)
p.show()
```

```python
s.barplot(x = months, y = sales)
p.show()
```

```python
s.scatterplot(x = months, y = sales)
p.show()
```

```python
s.histplot(sales)
p.show()
```

```python
courses = ['c', 'cpp', 'java', 'python', 'c', 'c', 'java', 'java', 'python']
s.countplot(x = courses)
p.show()
```

```python
import pandas as pd
data = {
    'Python':[80, 90, 70, 60],
    'Java':[20, 30, 50, 40],
    'SQL':[80, 30, 90, 40]
}
df = pd.DataFrame(data)
c =  df.corr()
s.heatmap(c)
p.show()
```
