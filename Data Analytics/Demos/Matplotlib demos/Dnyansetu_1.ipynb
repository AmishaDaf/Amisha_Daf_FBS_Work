# Barplot

```python
marks =[36, 78, 90, 81]
student = ["Ram", "Sham", "Sita", "Gita"]
pt.title("Marks of Students")
pt.xlabel("Students")
pt.ylabel("Marks")
pt.bar(student, marks)
pt.show()
```

```python
marks =[36, 78, 90, 81]
student = ["Ram", "Sham", "Sita", "Gita"]
pt.title("Marks of Students")
pt.xlabel("Students")
pt.ylabel("Marks")
pt.bar(student, marks, color="orange", edgecolor = "blue")
pt.show()
```

```python
pt.barh(student, marks, color="orange", edgecolor = "blue")
```

```python
marks =[36, 78, 90, 81]
student = ["Ram", "Sham", "Sita", "Gita"]
pt.title("Marks of Students")
pt.xlabel("Students")
pt.ylabel("Marks")
pt.bar(student, marks, color="orange", edgecolor = "blue")
pt.show()
pt.barh(student, marks, color="orange", edgecolor = "blue")
pt.show()
```
