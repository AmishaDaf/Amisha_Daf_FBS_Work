# Pie plot
#### 1. Showing percentage distribution
#### 2. Part to whole comparison

```python
subjects = ['Python', 'MySQL', 'DA', 'DS', 'WP']
marks = [90, 45, 50, 80, 70]
```

```python
plt.pie(marks)
```

```python
plt.pie(marks, labels = subjects)
```

```python
plt.pie(marks, labels = subjects, explode = [0.05, 0.02, 0.03, 0.06, 0.04])
```

```python
plt.pie(marks, labels = subjects, explode = [0.05, 0.02, 0.03, 0.06, 0.04], autopct='%0.2f%%')
plt.show()
```
