```python
import pandas as pd
```

```python
df1 = pd.DataFrame({'ID' : [101, 102, 103, 104],
                    'NAME' : ['Sumit', 'Priya', 'Anita', 'Supriya'],
                    'AGE' : [34, 56, 55,  29]})
```

```python
df1
```

```python
df2 = pd.DataFrame({'ID' : [104, 105, 106],
                    'NAME' : ['Rutu', 'Piyu', 'Pari'],
                    'AGE' : [39, 100, 49]})
df2
```

```python
pd.concat([df1,df2])
```

```python
df3 = pd.DataFrame({'ID' : [105, 106, 107],
                    'NAME' : ['Rutu', 'Piyu', 'Pari'],
                    'SAL' : [39, 100, 49]})
df3
```

```python
pd.concat([df1,df3])
```

```python
df4 = pd.DataFrame({'ADDRESS' : ['Pune', 'Mumbai', 'Chennai', 'Kashmir'],
                    'DEPT' : ['IT', 'DA', 'DS', 'AI/ML'],
                    'SAL' : [39000, 100000, 49000, 45000]})
df4
```

```python
df1
```

```python
pd.concat([df1, df4], axis = 1)
```

```python
emp = pd.DataFrame({'ID' : [101, 102, 103, 104],
                    'NAME' : ['Sumit', 'Priya', 'Anita', 'Supriya'],
                    'AGE' : [34, 56, 55,  29],
                     'DID': [1, 3, 2, 5]})
emp
```

```python
dept = pd.DataFrame({'DID':[1,2,3,4],
                     'DNAME': ['IT', 'Admin', 'HR', 'Sales']})
```

```python
dept
```

```python
pd.merge(emp, dept)
```

```python
dept2 = pd.DataFrame({'ID':[1,2,3,4],
                     'DNAME': ['IT', 'Admin', 'HR', 'Sales']})
dept2
```

```python
pd.merge(emp, dept2)
```

```python
pd.merge(emp, dept2, left_on = 'DID', right_on = 'ID')
```

```python
pd.merge(emp, dept, how='left')
```

```python
pd.merge(emp, dept, how='right')
```

```python
pd.merge(emp, dept, how='outer')
```
